```python
class Queue:
  def dequeue(self,qu):
    if(len(qu)) == 0:
      print("queue is empty")
    rem = qu.pop(0)
    print("the dequeued element is",rem)

  def enqueue(self, qu,data):
    qu.append(data)

  def peek(self,qu):
    print(qu[0])

  def size(self,qu):
    print(len(qu))

if __name__ == "__main__":
  var = Queue()
  li = [10,20,30,40,50]
  print(li)
  var.dequeue(li)
  print(li)
  var.enqueue(li,60)
  print(li)
  var.peek(li)
  var.size(li)
```
