```python
class Stack:
  def push(self, st, data):
    st.append(data)
  def pop(self,st):
    if len(st)>0:
      ans = st.pop()
      print("the popped elsement is :",ans)
    else:
      print("there is no element")

  def top(self,st):
    if len(st)==0:
      print("Stack is empty")
    else:
      print("top element is ",st[len(st)-1])

  def size(self,st):
    if len(st)==0:
      print("stack is empty")
    else:
      print(len(st))

if __name__ == "__main__":
  var = Stack()
  s= [10, 20, 30, 40, 50]
  print(s)
  var.push(s,60)
  print(s)

  var.pop(s)
  var.pop(s)
  print(s)
  var.top(s)
  var.size(s)



```
